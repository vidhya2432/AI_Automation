```python
##Task-1
##Read the CSV using Python
##How many students are there
import pandas as pd

df=pd.read_csv("/content/Students Performance Dataset (1).csv")
df.head()
```

```python
total_std=len(df)
print("Total Student:",total_std)
```

```python
#Task-2
# Find highest, lowest, average and department counts in Engineering students CS students
def find_highest(data):
  return data["Total_Score"].max()

def find_lowest(data):
  return data["Total_Score"].min()

def find_avg(data):
  return data["Total_Score"].mean()

def count_engineering(data):
  return (data["Department"] =="Engineering").sum()

def count_cs(data):
  return (data["Department"] =="CS").sum()

highest = find_highest(df)
lowest = find_lowest(df)
average = find_avg(df)
engineering_count = count_engineering(df)
cs_count = count_cs(df)

print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
print("Average Mark:", round(average, 2))
print("Engineering Students:", engineering_count)
print("CS Students:", cs_count)
```
